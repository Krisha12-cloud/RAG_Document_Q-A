## Step 4 — Configure: Choose Your PDF

Change `PDF_FILE` to whichever document you want to index.


```python
# ── CONFIGURE HERE ────────────────────────────────────────────────────────
PDF_FILE  = "ai_basics.pdf"     # Change to: climate_change_report.pdf
                                #            or: hr_policy_manual.pdf
                                #            or: path to your own PDF
INDEX_DIR = "faiss_index"       # Where the index will be saved

CHUNK_SIZE    = 500   # tokens per chunk  (increase for more context)
CHUNK_OVERLAP = 50    # overlap between chunks (prevents cutting sentences)
K_RETRIEVAL   = 3     # how many chunks to retrieve at query time

print(f" Will index: {PDF_FILE}")
print(f" Index will be saved to: {INDEX_DIR}/")
```

## Step 5 — Load Document

```python
# Auto-detect file type and load
if PDF_FILE.endswith(".pdf"):
    loader = PyPDFLoader(PDF_FILE)
else:
    loader = TextLoader(PDF_FILE)

documents = loader.load()
print(f" Loaded '{PDF_FILE}'")
print(f"   Pages / sections: {len(documents)}")
print(f"\n--- Preview of Page 1 (first 300 chars) ---")
print(documents[0].page_content[:300])
print(f"\nMetadata: {documents[0].metadata}")
```

##  Step 6 — Split into Chunks

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=CHUNK_SIZE,
    chunk_overlap=CHUNK_OVERLAP,
    separators=["\n\n", "\n", ".", " "]
    # tries double newline first, falls back to single newline, then period, then space
)

chunks = splitter.split_documents(documents)

print(f" Created {len(chunks)} chunks from {len(documents)} page(s)")
print(f"   Chunk size: {CHUNK_SIZE} tokens | Overlap: {CHUNK_OVERLAP} tokens")
print(f"\n--- Chunk 1 ---")
print(textwrap.fill(chunks[0].page_content, width=80))
print(f"\n--- Chunk 2 ---")
print(textwrap.fill(chunks[1].page_content, width=80))
```

##  Step 7 — Load Embedding Model

We use `all-MiniLM-L6-v2` from HuggingFace 


```python
print(" Loading embedding model (downloads ~90 MB on first run)...")

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# Quick test — embed one sentence
test_vector = embeddings.embed_query("test sentence")
print(f" Embedding model loaded!")
print(f"   Vector dimensions: {len(test_vector)}")
print(f"   Sample values: {test_vector[:5]}")
```

## Step 8 — Build & Save FAISS Index

```python
print(" Embedding all chunks and building FAISS index...")

vectorstore = FAISS.from_documents(chunks, embeddings)
vectorstore.save_local(INDEX_DIR)

print(f" FAISS index saved to '{INDEX_DIR}/'")
print(f"   Total vectors stored: {vectorstore.index.ntotal}")
print(f"   Vector dimensions:    {vectorstore.index.d}")
print(f"\n Indexing complete! Open query.ipynb to start asking questions.")
```
